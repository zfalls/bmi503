## Creating arrays with functions

Above, we manually built arrays, but this is not feasible with very large sets of data. We can automate the creation of various arrays using a set of functions.

The first such function is `.arange()` which functions similarly to Python's `range()` except the output is a NumPy array.

```python
np.arange(5)
```

```python
np.arange(1,5,2)
```

`.arange()` is great at creating an array with evenly spaced values based on the start, end, and step. But what if we what a certain number of items in a list and let the computer determine what those values are? For this we can use `.linspace()`. 

Here we provide the start, end, and number of items as arguments.

```python
np.linspace(0,4,5)
```

```python
np.linspace(0,1,5)
```

`.arange()` and `.linspace()` both generate 1D arrays by default. But we can apply another function to these arrays (or any array) to manipulate the size of each dimension. To this end, we use the `.reshape()` method.

For this method, we pass the new dimensions as a tuple.

```python
a = np.arange(6)
a
```

```python
a.reshape((2,3))
```

We can also infer the size of a dimension by passing `-1`.

```python
a.reshape((2,-1))
```

To flatten an array (i.e., undo a reshape and make a 1D array) you can use the `.ravel()` method.

```python
a.ravel()
```

We can also build 2D or larger arrays with various properties using specific functions.

The first of these is `.ones()`, which, as the name suggests, builds an matrix of ones. Yo.median()`, and `.std()`u just need to pass the dimensions of the matrix into the function as a tuple.

```python
np.ones((3, 3))
```

The opposite functions, `.zeros()` provides a null matrix of all zeros.

All you need to pass is a tuple of the array shape.

```python
np.zeros((3,3))
```

Another function to create arrays is the `.eye()` function, which build a square diagonal matrix, i.e. an identity matrix.

All you need to pass is the number of values in the diagonal (or the dimension of the array).

```python
np.eye(3)
```

Lastly, we can build a square matrix with a defined diagonal using `.diag()`. All we need to pass is a numpy array of the desired diagonal.

```python
np.diag(np.array([1,2,3]))
```

## Arithmetic operations

We can also manipulate arrays by applying math operators to them. The typical operators from Python work with arrays, but you must be aware of the rules. 

Single values (scalars) apply elementwise to the array. This means that whatever operation you apply to the array with a scalr will apply to every item in the array.

```python
a = np.array([0,1,2])
a
```

```python
a + 1
```

```python
a * 4
```

Just a reminder, we do not reassign the array when we apply these operators. It merely applies them to the view of the array and returns the modified array. We would need to reassign or create a new variable to keep the changes.

```python
b = a * 4
b
```

```python
a
```

When applying math operators between arrays, it remains elementwise.

```python
a = np.array([0,1,2])
b = np.array([3,4,5])
```

```python
a+b
```

Using the `*` operator does not multiply the matrices, but multiplies the corresponding elements in the matrix. To do matrix multiplication, you must use the `@` operator. 

```python
a = np.arange(9)
a = a.reshape((3,3))
```

```python
a*a
```

```python
a@a
```

So when we apply matrix multiplication between an identity matrix and our array when should return the same array.

```python
i = np.eye(3)
```

```python
a@i
```

What happens when you have two arrays that do not have the same shape or dimensions? This is when a concept called broadcasting occurs in NumPy.

```python
a = np.arange(9)
a = a.reshape((3,3))

b = np.array([3,4,5])
```

```python
a
```

```python
b
```

```python
a+b
```

As you can see, we had a 3x3 matrix summed with a 1D array of length 3. What NumPy does is modifies the arrays so they have common dimesions and then applies the operation across the arrays. 

However, the arrays must have a dimension in common in order to properly apply the operation.

```python
a = np.array([[0,1,2,3]])
b = np.array([3,4,5])
a+b
```

We can also use comparison operators...

```python
a = np.array([0,1,2])
b = np.array([0,1,3])
```

```python
a == b
```

```python
a < b
```

And, of course, since we are talking about matrices, we can get the transpose of an array using `.T`.

```python
a = np.array([[0,1,2],[3,4,5]])
```

```python
a.T
```

## Randomize

A very useful submodule in numpy is random. The functions can be called using `np.random`.

Three functions that are really useful are `.rand()` which will create an array populated with random values between 0 and 1. All you have to pass are the dimensions of the array (not as a tuple this time). 

```python
np.random.rand(3,3)
```

If you notice, each time you run the above cell it gives a different set of values each time. In most cases we will want to be deterministic every time we want to generate random values. To do this, we must set a random seed, which is the starting point for our random generator. THis is done using the `.seed()` function. Pass any int you want.

```python
np.random.seed(42)
np.random.rand(3,3)
```

Lastly, we were sorting arrays previously. But, in some cases, you may want to randomize your data. In that case, you can use the `.shuffle()` function. 

```python
x = np.arange(10)
x
```

```python
np.random.shuffle(x)
x
```

Don't forget to set the random seed if you want it shuffled the same every time.

```python
np.random.seed(42)

x = np.arange(10)
np.random.shuffle(x)
x
```
